```python
import matplotlib.pyplot as plt
```

```python
track_avg = eceboard2.groupby('Track')['Average'].mean().reset_index()
gender_avg = eceboard2.groupby('Gender')['Average'].mean().reset_index()
hometown_avg = eceboard2.groupby('Hometown')['Average'].mean().reset_index()
```

```python
display(track_avg)
display(gender_avg)
display(hometown_avg)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].bar(track_avg['Track'], track_avg['Average'])
axes[0].set_xlabel('Track')
axes[0].set_ylabel('Mean Average')
axes[1].bar(gender_avg['Gender'], gender_avg['Average'])
axes[1].set_xlabel('Gender')
axes[1].set_ylabel('Mean Average')
axes[2].bar(hometown_avg['Hometown'], hometown_avg['Average'])
axes[2].set_xlabel('Hometown')
axes[2].set_ylabel('Mean Average')
fig.text(0,0, 'The track with the highest mean average of the three is Communication.',)

fig.text(0,-0.05, 'The Gender with the highest mean average is Male.',)

fig.text(0,-0.10, 'The Hometown with the highest mean average of the the three islands is Luzon.',)
fig.tight_layout()
plt.show()

```
